### Campo de Galois GF(2^m)

A) m= 8  ⇒ GF (2^8)

p(x)= X^8 + X^4 + X^3 + X^2 + 1 ⇒ bin= 1|00011101

B) m=4  ⇒ GF (2^4)

p(x)= X^4 + X + 1 ⇒ bin= 1|0011

```python
#Instancia de la clase

#Polinomio primitivo: P(x) = X^8+X^4+1 -> bin: 1|00011101 -> dec=17 (sin el término X^m)

pol_pri_gf8_bin="00011101"
primitivo_gf8=int(pol_pri_gf8_bin,2)

gf8=GF(m=8,p_x=primitivo_gf8) 
print (f"---------------Cuerpo de Galois (2^{gf8.m})----------------")
print(f"Polinomio primitivo: \n\tP(x) = X^8+X^4+X^3+X^2+1 \n\tp={gf8.p_x} (binario:{bin(gf8.p_x)}) sin el termino x^{gf8.m}")
print(f"Elementos van de 0 a {gf8.mask}\n")



#Polinomio primitivo: P(x)=X^4+X+1 -> bin:1|0011 -> dec=3 (sin el término X^m)

pol_pri_gf4_bin="0011"
primitivo_gf4=int(pol_pri_gf4_bin,2)

gf4=GF(m=4,p_x=3) 
print (f"---------------Cuerpo de Galois (2^{gf4.m})----------------")
print (f"Polinomio primitivo: \n\tP(x)=X^4+X+1 \n\tp={gf4.p_x} (binario:{bin(gf4.p_x)}) sin el termino x^{gf4.m}")
print(f"Elementos van de 0 a {gf4.mask}")

```

```python
#Verifico que la función orden_de_elemento() me devuelva el orden de los elementos
print("\nVerificación de que la función orden_de_elemento() me devuelva el orden de los elementos:\n")

alfa_5_bin="0110" #alfa^5
alfa_5=int(alfa_5_bin,2)

print(f"Orden de elemento α^5=α^2 +α : {gf4.orden_de_elemento(alfa_5)}")
```

```python
#Verificación de que alfa es de orden 2^m-1 en ambos cuerpos de Galois

print("Verificación de que α es de orden 2^m-1 en ambos cuerpos de Galois:\n")
print(f"1) Para el cuerpo de Galois GF(2^{gf8.m}):")
print("Orden de elemento α:", gf8.orden_de_elemento(2))


print(f"\n2) Para el cuerpo de Galois GF(2^{gf4.m}):")
print("Orden de elemento α: ", gf4.orden_de_elemento(2))
```

```python
F=[1,13,1]  
G=[15,3,1]   
H=[3,2,1]
K=[1,1,0,1,1]


poly1=GFPoly(gf4,F)
poly2=GFPoly(gf4,G)
poly3=GFPoly(gf4,H)
poly4=GFPoly(gf4,K)
```
